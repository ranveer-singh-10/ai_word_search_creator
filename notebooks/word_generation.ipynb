```python
import json
from dotenv import load_dotenv
import anthropic

load_dotenv()
client = anthropic.Anthropic()
```

```python
def generate_words(category, difficulty, count=10):
    system_prompt = (
        "You generate words for word search puzzles. Easy difficulty means common words of 3–6 letters, and hard means less common words of 7–12 letters, medium means a mix of both"
        "Respond with JSON only and include a small description for each word, no other text, in this format: "
        '{"words": [{"word": "COMET", "description": "An icy body that develops a tail near the Sun"},    {"word": "ORBIT", "description": "The curved path of one object around another"}]}'
    )

    user_prompt = (
        f"Category: {category}\n"
        f"Difficulty: {difficulty}\n"
        f"Give me {count} words. Single words only, letters A-Z, no spaces or hyphens or special characters"
    )

    response = client.messages.create(
        model="claude-haiku-5-5",
        max_tokens=1000,
        system=system_prompt,
        messages=[{"role": "user", "content": user_prompt}],
    )

    text = next(block.text for block in response.content if block.type == "text")
    print(response.usage)
    return json.loads(text)["words"]
```

```python
words = generate_words("Space", "easy")
print(words)
```

```python
def validate_words(words, max_length=12):
    clean = []
    for word in words:
        word_word = word["word"].upper().strip()
        if not word_word.isalpha():
            continue          # skip anything with spaces, hyphens, numbers
        if len(word_word) > max_length:
            continue          # won't fit in the grid
        if word_word in clean:
            continue          # duplicate
        clean.append(word_word)

    return clean

print(validate_words(words))
```
