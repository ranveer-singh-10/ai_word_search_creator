```python
import json
from dotenv import load_dotenv
import anthropic

from wordsearch.grid import place_word_in_grid

load_dotenv()
client = anthropic.Anthropic()
```

```python
def generate_words(category, difficulty, count=10):
    system_prompt = (
        "You generate words for word search puzzles. Easy difficulty means common words of 3–6 letters, and hard means less common words of 7–12 letters, medium means a mix of both"
        "Respond with JSON only and include a small description for each word, no other text, in this format: "
        '{"words": [{"word": "COMET", "description": "An icy body that develops a tail near the Sun"},    {"word": "ORBIT", "description": "The curved path of one object around another"}]}'
    )

    user_prompt = (
        f"Category: {category}\n"
        f"Difficulty: {difficulty}\n"
        f"Give me {count} words. Single words only, letters A-Z, no spaces or hyphens or special characters"
    )

    response = client.messages.create(
        model="claude-haiku-5-5",
        max_tokens=1000,
        system=system_prompt,
        messages=[{"role": "user", "content": user_prompt}],
    )

    text = next(block.text for block in response.content if block.type == "text")
    print(response.usage)
    return json.loads(text)["words"]
```

```python
words = generate_words("Space", "easy")
print(words)
```

```python
def validate_words(words, max_length=12):
    clean = []
    for word in words:
        word_word = word["word"].upper().strip()
        if not word_word.isalpha():
            continue          # skip anything with spaces, hyphens, numbers
        if len(word_word) > max_length:
            continue          # won't fit in the grid
        if word_word in clean:
            continue          # duplicate
        clean.append(word_word)

    return clean

print(validate_words(words))
```

```python
def word_fits_from(grid, word, start, direction):
    """True if word reads from start, stepping in direction, with every letter matching."""
    position = start
    for i, letter in enumerate(word):
        row, col = position
        if grid[row][col] != letter:
            return False
        if i < len(word) - 1:          # only move if there's another letter to check
            position = next_cell(grid, position, direction)
            if position is None:       # ran off the edge before the word finished
                return False
    return True


def check_word_in_grid(grid, word, directions):
    """True if word appears anywhere in grid in any of the given directions."""
    for row in range(len(grid)):
        for col in range(len(grid[0])):
            for direction in directions:
                if word_fits_from(grid, word, (row, col), direction):
                    return True
    return False
```

```python
# Each direction is (row_step, col_step).
DIRECTIONS = {
    "right":      (0, 1),
    "left":       (0, -1),
    "down":       (1, 0),
    "up":         (-1, 0),
    "down_right": (1, 1),
    "down_left":  (1, -1),
    "up_right":   (-1, 1),
    "up_left":    (-1, -1),
}

# Which directions each difficulty allows (from the project brief).
ALLOWED_DIRECTIONS = {
    "easy":   ["right", "down"],
    "medium": ["right", "down", "down_right", "up_right"],
    "hard":   list(DIRECTIONS),
}

def next_cell(grid, grid_position, direction):
    row, col = grid_position
    row_step, col_step = direction
    next_row = row + row_step
    next_col = col + col_step
    if 0 <= next_row < len(grid) and 0 <= next_col < len(grid[0]):
        return (next_row, next_col)
    return None
```
